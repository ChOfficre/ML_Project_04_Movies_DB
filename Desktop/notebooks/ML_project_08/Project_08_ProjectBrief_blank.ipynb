## Machine Learning - Predicting House Values (optional)

Based on the Explanatory Data Analysis, <br>
- select an appropriate ML Model (e.g. Linear Regression or RandomForest Regression)
- split into Train and Test Set
- prepare/engineer Features
- Fit the Model (Train Set)
- Test the Model (Test Set)

```python
labels = df.median_house_value.copy()
labels
```

```python
features = df.drop(columns=["median_house_value"])
features
```

```python
features.info()
```

Fm all data type in features, take only numerical data type

```python
features.select_dtypes("float")
```

Compute the z score of each value in the sample, relative to the sample mean and standard deviation.

```python
import scipy.stats as stats
```

```python
feat1 = features.select_dtypes("float").apply(lambda x: stats.zscore(x))
feat1
```

```python
pd.options.display.float_format = "{:.2f}".format
```

For all columns create mean and standat deviation

```python
feat1.agg(["mean", "std"])
```

Changing objet data to numerical data before start ML models.
All data must be numeral

```python
features.ocean_proximity
```

```python
features.ocean_proximity.value_counts()
```

```python
dummies = pd.get_dummies(features.ocean_proximity)
dummies
```

Concatinate all df to one + add column income_cat 

```python
features = pd.concat([feat1, dummies, df.income_cat], axis=1)
features
```

## Splitting the Data into Train and Test Set

```python
features
```

```python
test_size = 0.2
```

```python
X_test = features.sample(frac=test_size, random_state=123)
X_test
```

Check that equal parameters (Income_cat) are in test set

```python
features.income_cat.value_counts(normalize=True)
```

```python
X_test.income_cat.value_counts(normalize=True)
```

Create Intex for X_test values, creat Train set without X_test index values. ~ == not

```python
X_test.index
```

```python
X_train = features.loc[~features.index.isin(X_test.index)].copy()
X_train
```

```python
X_train.income_cat.value_counts(normalize=True)
```

Shafl all row in X_train in random way

```python
X_train = X_train.sample(frac=1, random_state=123)
X_train
```

```python
X_test.drop(columns=["income_cat"], inplace=True)
X_train.drop(columns=["income_cat"], inplace=True)
```

```python
y_train = labels.loc[X_train.index]
y_test = labels.loc[X_test.index]
```

```python
y_train
```

## Training the ML Model (Random Forest Regressor)

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
forest_reg = RandomForestRegressor(random_state=42, n_estimators=500, max_features="sqrt", max_depth=75, min_samples_split=2)
```

```python
forest_reg.fit(X_train, y_train)
```

As closer to 1 is beter

```python
forest_reg.score(X_train, y_train)
```

An other option. as closer to 0 as beter

```python
from sklearn.metrics import mean_squared_error
```

```python
pred = forest_reg.predict(X_train)
pred
```

```python
forest_mse = mean_squared_error(y_train, pred)
forest_rmse = np.sqrt(forest_mse)
forest_rmse
```

## Evaluating the Model on the Test Set

```python
forest_reg
```

```python
forest_reg.score(X_test, y_test)
```

```python
pred = forest_reg.predict(X_test)
pred
```

```python
forest_mse = mean_squared_error(y_test, pred)
forest_rmse = np.sqrt(forest_mse)
forest_rmse
```

House Prices compare: Ture and Predict values

```python
comp = pd.DataFrame(data={"True_v": y_test, "Pred_v": pred})
comp
```

Compare diffirent of prices in absolut value

```python
ae = comp.True_v.sub(comp.Pred_v).abs()
ae
```

Mean value of compered prices

```python
mae = ae.mean()
mae
```
